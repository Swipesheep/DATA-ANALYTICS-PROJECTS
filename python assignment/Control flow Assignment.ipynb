```python
def check_leap_year(year):
        if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
            print(f"{year} is a Leap Year.")
        else:
            print(f"{year} is NOT a Leap Year.")
   
print("1. LEAP YEAR CHECKER")
print("=" * 50)
year_input = int(input("Enter your Year"))
check_leap_year(year_input)
```

```python
def largest_of_three(a, b, c):
    if a >= b:
        if a >= c:
            largest = a
        else:
            largest = c
    else:
        if b >= c:
            largest = b
        else:
            largest = c
    print(f"The largest of {a}, {b}, {c} is: {largest}")

print("\n" + "=" * 50)
print("2. LARGEST OF THREE NUMBERS")
print("=" * 50)
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
num3 = int(input("Enter third number: "))
largest_of_three(num1, num2, num3)
```

```python
#4.	Build a program that calculates and assigns grades to students based on their scores. Use if-else statements to determine the grade.
```

```python
def assign_grade(score):
    if score >= 90:
        grade = 'A'
    elif score >= 80:
        grade = 'B'
    elif score >= 70:
        grade = 'C'
    elif score >= 60:
        grade = 'D'
    else:
        grade = 'F'
    return grade  # Return the grade so we can use it in the loop

print("\n" + "=" * 50)
print("4. MULTI-SUBJECT GRADE CALCULATOR")
print("=" * 50)
user_input =input("Enter 5 subject scores separated by commas (e.g. 95, 82, 70, 64, 45): ")

try:
    score_list = [int(s.strip()) for s in user_input.split(",")]

    # 3. Check if exactly 5 scores were provided
    if len(score_list) == 5:
        print("-" * 30)
        for i, score in enumerate(score_list, 1):
            grade_result = assign_grade(score)
            print(f"Subject {i} | Score: {score} --> Grade: {grade_result}")
        print("-" * 30)
    else:
        print(f"Error: You entered {len(score_list)} scores. Please provide exactly 5.")

except ValueError:
    print("Error: Please enter valid numbers separated by commas.")
```

```python
#5.	Write a program that counts the number of vowels in a given string and displays the count for each vowel (a, e, i, o, u).
```

```python
def count_vowels(text):
    vowels = 'aeiou'
    counts = {v: 0 for v in vowels}
    for char in text.lower():
        if char in counts:
            counts[char] += 1
    print(f"Text: '{text}'")
    for vowel, count in counts.items():
        print(f"  '{vowel}': {count}")
    print(f"  Total vowels: {sum(counts.values())}")

print("\n" + "=" * 50)
print("5. VOWEL COUNTER")
print("=" * 50)
user_input =input("Enter your Words: ")
count_vowels(user_input)
```

```python
#6.	Create a program to generate and display a list of prime numbers within a specified range using a for loop.
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

def primes_in_range(start, end):
    primes = [n for n in range(start, end + 1) if is_prime(n)]
    print(f"Prime numbers between {start} and {end}:")
    prime_count = len(primes)
    print(primes)
    print("Count Of Prime Nummbers:",prime_count)
    

print("\n" + "=" * 50)
print("6. PRIME NUMBERS IN RANGE")
print("=" * 50)
starting_number =int(input("Enter your starting number: "))
ending_number =int(input("Enter your ending number: "))
primes_in_range(starting_number,ending_number)
```

```python
#7.	Write a program that generates the first n terms of the Fibonacci sequence using a for loop.
```

```python
def fibonacci(n):
    sequence = []
    a, b = 0, 1
    for i in range(n):
        sequence.append(a)
        a, b = b, a + b
    print(f"First {n} Fibonacci numbers: {sequence}")

print("\n" + "=" * 50)
print("7. FIBONACCI SEQUENCE")
print("=" * 50)
First_Fibonacci_numbers = int(input("Enter your number: "))
fibonacci(First_Fibonacci_numbers)
```

```python
#10. Write a program that checks if a given word or phrase is a palindrome (reads the same backward and forward) using a for loop.
```

```python
def is_palindrome(text):
    cleaned = ''.join(char.lower() for char in text if char.isalnum())
    reversed_text = ''
    for char in cleaned:
        reversed_text = char + reversed_text
    result = cleaned == reversed_text
    print(f"'{text}' is {'a palindrome' if result else 'NOT a palindrome'}")

print("\n" + "=" * 50)
print("10. PALINDROME CHECKER")
print("=" * 50)
is_palindrome("racecar")
is_palindrome("hello")
is_palindrome("A man a plan a canal Panama")
```

```python
#11. Create a program that reverses a given number using a while loop. For example, 12345 should become 54321.
```

```python
def reverse_number(n):
    original = n
    reversed_n = 0
    while n > 0:
        digit = n % 10
        reversed_n = reversed_n * 10 + digit
        n //= 10
    print(f"Original: {original} --> Reversed: {reversed_n}")

print("\n" + "=" * 50)
print("11. REVERSE A NUMBER")
print("=" * 50)
Put_numbers_to_reverse = int(input("Enter your number to be Reversed: "))
reverse_number(Put_numbers_to_reverse)
```

```python
#12.	Implement a program that generates a random password of a specified length using a while loop. Include a mix of letters, numbers, and special characters.
```

```python
import random
import string
def generate_password(length):
    characters = string.ascii_letters + string.digits + string.punctuation
    password = ''
    i = 0
    while i < length:
        password += random.choice(characters)
        i += 1
    print(f"Generated Password ({length} chars): {password}")

print("\n" + "=" * 50)
print("12. RANDOM PASSWORD GENERATOR")
print("=" * 50)
generate_password(12)
```

```python
#13.	Write a program that generates and displays the Collatz sequence for a given number. The sequence should end when it reaches 1.
```

```python
def collatz_sequence(n):
    sequence = [n]
    while n != 1:
        if n % 2 == 0:
            n //= 2
        else:
            n = 3 * n + 1
        sequence.append(n)
    print(f"Collatz sequence: {sequence}")
    print(f"Length: {len(sequence)} steps")

print("\n" + "=" * 50)
print("13. COLLATZ SEQUENCE")
print("=" * 50)
collatz_sequence(8)
collatz_sequence(18)
```

```python
#14.Build a number guessing game where the program selects a random number, and the user has to guess it. Use a while loop to continue until the user guesses correctly
```

```python
import random

def number_guessing_game():
    target = random.randint(1, 100)
    attempts = 0
    
    print("Number Guessing Game! (I've picked a number between 1–100)")
    
    while True:
        try:
            # Move the input INSIDE the loop so you can guess multiple times
            guess = int(input("\nEnter your guess: "))
            attempts += 1
            
            if guess < target:
                print(f"  Guess {attempts}: {guess} → Too Low!")
            elif guess > target:
                print(f"  Guess {attempts}: {guess} → Too High!")
            else:
                print(f"  Correct! 🎉 You found it in {attempts} attempts.")
                break # Exit the loop when you win
        except ValueError:
            print("Please enter a valid whole number!")

print("\n" + "=" * 50)
print("14. NUMBER GUESSING GAME")
print("=" * 50)

# Start the game (no arguments needed)
number_guessing_game()
```

```python
#15.Create a program to find the greatest common divisor (GCD) of two numbers using the Euclidean algorithm with a while loop.
```

```python
def gcd(a, b):
    original_a, original_b = a, b
    while b != 0:
        a, b = b, a % b
    print(f"GCD({original_a}, {original_b}) = {a}")

print("\n" + "=" * 50)
print("15. GCD (EUCLIDEAN ALGORITHM)")
print("=" * 50)
gcd(48, 18)
gcd(100, 75)
```

```python
#16.Continue and Break Statements:
#17.Write a program that uses the continue statement to find and display the first n prime numbers.
```

```python
def first_n_primes(n):
    primes = []
    candidate = 2
    while len(primes) < n:
        is_p = True
        for i in range(2, int(math.sqrt(candidate)) + 1):
            if candidate % i == 0:
                is_p = False
                break
        if not is_p:
            candidate += 1
            continue   # <-- continue used here
        primes.append(candidate)
        candidate += 1
    print(f"First {n} prime numbers: {primes}")

print("\n" + "=" * 50)
print("17. FIRST N PRIMES (continue statement)")
print("=" * 50)
first_n_primes(10)
first_n_primes(15)
```

```python
#18.Create a program that finds and displays all palindrome words in a given sentence using the break statement to stop after the first one.
```

```python
def first_palindrome_word(sentence):
    words = sentence.split()
    print(f"Sentence: '{sentence}'")
    for word in words:
        cleaned = word.strip(".,!?").lower()
        if cleaned == cleaned[::-1] and len(cleaned) > 1:
            print(f"First palindrome word found: '{word}'")
            break   # <-- break used here
    else:
        print("No palindrome words found.")

print("\n" + "=" * 50)
print("18. FIRST PALINDROME WORD (break statement)")
print("=" * 50)
first_palindrome_word("The Man is Calling for SOS ")
first_palindrome_word("Hello world Python")
```

```python
#19.Build a program that finds and displays all the factors of a given number, using the break statement to stop once a factor exceeds the square root of the number.
```

```python
def find_factors(n):
    factors = []
    print(f"Factors of {n}:")
    for i in range(1, n + 1):
        if i > math.sqrt(n) and i not in factors:
            break   # <-- break used here
        if n % i == 0:
            factors.append(i)
            if i != n // i:
                factors.append(n // i)
    factors.sort()
    print(f"  {factors}")

print("\n" + "=" * 50)
print("19. FACTORS WITH BREAK")
print("=" * 50)
find_factors(36)
find_factors(100)
```

```python
# 21.Write a program that checks the strength of a password provided by the user. Use the continue statement to validate various criteria, such as length, uppercase and lowercase letters, and special characters.
```

```python
def check_password_strength(password):
    print(f"\nChecking password: '{password}'")
    criteria = {
        "Length >= 8"      : len(password) >= 8,
        "Uppercase letter" : any(c.isupper() for c in password),
        "Lowercase letter" : any(c.islower() for c in password),
        "Digit"            : any(c.isdigit() for c in password),
        "Special character": any(c in string.punctuation for c in password),
    }

    passed = 0
    for criterion, met in criteria.items():
        if not met:
            print(f"  ✗ Missing: {criterion}")
            continue          # <-- continue skips counting unmet criteria
        print(f"  ✓ Passed: {criterion}")
        passed += 1

    strength = ["Very Weak", "Weak", "Fair", "Strong", "Very Strong"][passed - 1] if passed else "Very Weak"
    print(f"  Password Strength: {strength} ({passed}/{len(criteria)} criteria met)")

print("\n" + "=" * 50)
print("21. PASSWORD STRENGTH CHECKER (continue statement)")
print("=" * 50)
Checking_password = (input("Enter your Password: "))
check_password_strength(Checking_password)
```
